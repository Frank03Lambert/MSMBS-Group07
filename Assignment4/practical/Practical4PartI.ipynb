```python
def calculate_distance_matrix(model: np.ndarray) -> np.ndarray:
    """
    Calculate the distance matrix for a given model.

    Parameters:
    model (np.ndarray): The input model as a 2D numpy array.

    Returns:
    np.ndarray: The distance matrix as a 2D numpy array.
    """
    num_rows = model.shape[0]
    distance_matrix = np.zeros((num_rows, num_rows))

    for i in range(num_rows):
        for j in range(num_rows):
            distance_matrix[i, j] = np.linalg.norm(model[i] - model[j])

    return distance_matrix
```

```python
# Print distance matrices for the models
rdm_model1 = calculate_distance_matrix(model1)
rdm_model2 = calculate_distance_matrix(model2)
rdm_brain = calculate_distance_matrix(brain)
print("Distance Matrix for Model 1:\n", rdm_model1)
print("Distance Matrix for Model 2:\n", rdm_model2)
print("Distance Matrix for Brain:\n", rdm_brain)
```

```python
# Calculate similarity between the models and the brain using the Pearson Correlation Coefficient
def calculate_similarity(rdm1: np.ndarray, rdm2: np.ndarray) -> float:
    """
    Calculate the similarity between two distance matrices using the Pearson Correlation Coefficient.

    Parameters:
    rdm1 (np.ndarray): The first distance matrix.
    rdm2 (np.ndarray): The second distance matrix.

    Returns:
    float: The Pearson Correlation Coefficient between the two distance matrices.
    """
    # Flatten the upper triangular part of the distance matrices to avoid redundancy
    rdm1_flat = rdm1[np.triu_indices_from(rdm1, k=1)]
    rdm2_flat = rdm2[np.triu_indices_from(rdm2, k=1)]

    # Calculate the Pearson Correlation Coefficient.
    correlation_coefficient = np.corrcoef(rdm1_flat, rdm2_flat)[0, 1]
    
    return correlation_coefficient
```

```python
PCC_model1_brain = calculate_similarity(rdm_model1, rdm_brain)
PCC_model2_brain = calculate_similarity(rdm_model2, rdm_brain)
print(f"Pearson Correlation Coefficient between Model 1 and Brain: {PCC_model1_brain:.4f}")
print(f"Pearson Correlation Coefficient between Model 2 and Brain: {PCC_model2_brain:.4f}")
```

Model 1 is the better fit, because it has a very high Pearson Correlagtion Coefficient (whereas the Pearson Correlation Coefficient of model 2 is way lower). If you look at the matrices, you can also see why this is the case, because if you look at where the high and low values are located, the matrices of model 1 and the brain look quite similar. In contrast, the matrix for model 2 has high values at locations where the brain has low values, for example.
